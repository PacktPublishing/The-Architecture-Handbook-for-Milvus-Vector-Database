## Decision Framework Functions

```python
def assess_requirements():
    """Interactive assessment of multi-tenancy requirements"""
    print("📋 Multi-Tenancy Requirements Assessment")
    print("=" * 50)

    # For demo purposes, we'll use predefined scenarios
    # In practice, these would be user inputs

    scenarios = {
        "startup_saas": {
            "tenant_count": 50,
            "growth_rate": "high",
            "isolation_requirement": "medium",
            "budget_constraint": "high",
            "compliance_requirement": "low",
            "performance_tier": "standard",
            "operational_complexity": "low",
        },
        "enterprise_platform": {
            "tenant_count": 25,
            "growth_rate": "medium",
            "isolation_requirement": "high",
            "budget_constraint": "low",
            "compliance_requirement": "high",
            "performance_tier": "premium",
            "operational_complexity": "medium",
        },
        "high_scale_consumer": {
            "tenant_count": 2000,
            "growth_rate": "very_high",
            "isolation_requirement": "medium",
            "budget_constraint": "high",
            "compliance_requirement": "medium",
            "performance_tier": "mixed",
            "operational_complexity": "low",
        },
    }

    return scenarios


def recommend_strategy(requirements):
    """Recommend multi-tenancy strategy based on requirements"""
    tenant_count = requirements["tenant_count"]
    isolation = requirements["isolation_requirement"]
    budget = requirements["budget_constraint"]
    compliance = requirements["compliance_requirement"]

    recommendations = []

    # Decision logic
    if compliance == "high" and isolation == "high":
        recommendations.append(
            {
                "strategy": "Database-Level Isolation",
                "priority": "primary",
                "reason": "Maximum security and compliance requirements",
            }
        )

    if tenant_count > 100 and budget == "high":
        recommendations.append(
            {
                "strategy": "Partition Key-Based",
                "priority": "primary",
                "reason": "High scalability with cost efficiency",
            }
        )

    if tenant_count <= 100 and budget == "high":
        recommendations.append(
            {
                "strategy": "Manual Partition-Based",
                "priority": "primary",
                "reason": "Balanced approach for moderate scale",
            }
        )

    if requirements["performance_tier"] in ["premium", "mixed"]:
        recommendations.append(
            {
                "strategy": "Resource Groups",
                "priority": "secondary",
                "reason": "Performance isolation for premium tiers",
            }
        )

    return recommendations


# Assess different scenarios
scenarios = assess_requirements()

for scenario_name, requirements in scenarios.items():
    print(f"\n📊 Scenario: {scenario_name.replace('_', ' ').title()}")
    print(f"   Tenants: {requirements['tenant_count']}")
    print(f"   Isolation: {requirements['isolation_requirement']}")
    print(f"   Budget constraint: {requirements['budget_constraint']}")
    print(f"   Compliance: {requirements['compliance_requirement']}")

    recommendations = recommend_strategy(requirements)

    print("   \n🎯 Recommendations:")
    for rec in recommendations:
        priority = "🥇" if rec["priority"] == "primary" else "🥈"
        print(f"   {priority} {rec['strategy']}: {rec['reason']}")
```

## Hybrid Strategy Implementation

```python
# Demonstrate a hybrid approach combining multiple strategies
def implement_hybrid_strategy():
    """Implement a hybrid multi-tenancy approach"""
    print("🔧 Implementing Hybrid Multi-Tenancy Strategy")
    print("=" * 50)
    print("Scenario: SaaS platform with different customer tiers")
    print("- Enterprise customers: Database isolation")
    print("- Standard customers: Partition-based sharing")
    print("- Basic customers: Shared partitions with filtering")

    # Define hybrid architecture
    hybrid_config = {
        "enterprise_tenants": {
            "strategy": "database",
            "tenants": ["enterprise_corp_a", "enterprise_corp_b"],
            "collections_per_tenant": ["products", "analytics", "users"],
        },
        "standard_tenants": {
            "strategy": "partition_key",
            "collection": "standard_shared_products",
            "tenants": [
                "standard_company_1",
                "standard_company_2",
                "standard_company_3",
            ],
        },
        "basic_tenants": {
            "strategy": "manual_partition",
            "collection": "basic_shared_catalog",
            "tenants": ["basic_user_1", "basic_user_2", "basic_user_3", "basic_user_4"],
        },
    }

    return hybrid_config


def setup_enterprise_tier(tenant_list):
    """Setup database-level isolation for enterprise customers"""
    print("\n🏢 Setting up Enterprise Tier (Database Isolation):")

    for tenant_id in tenant_list:
        db_name = f"db_{tenant_id}"

        try:
            # Create dedicated database
            client.create_database(db_name)
            print(f"  ✓ Created database: {db_name}")

            # Switch to tenant database and create collections
            client.using_database(db_name)

            # Create a sample collection
            schema = client.create_schema(auto_id=True)
            schema.add_field("id", DataType.INT64, is_primary=True)
            schema.add_field("vector", DataType.FLOAT_VECTOR, dim=384)
            schema.add_field("content", DataType.VARCHAR, max_length=500)

            client.create_collection(
                collection_name="enterprise_products", schema=schema
            )
            print(f"  ✓ Created collection for {tenant_id}")

        except Exception as e:
            print(f"  ⚠️  {tenant_id}: {str(e)[:50]}...")

    # Switch back to default database
    client.using_database("default")


def setup_standard_tier(tenant_list):
    """Setup partition key-based isolation for standard customers"""
    print("\n🏬 Setting up Standard Tier (Partition Key Isolation):")

    collection_name = "standard_products_pk"

    # Create collection with partition key
    schema = client.create_schema(auto_id=True)
    schema.add_field("id", DataType.INT64, is_primary=True)
    schema.add_field(
        "tenant_id", DataType.VARCHAR, max_length=50, is_partition_key=True
    )
    schema.add_field("vector", DataType.FLOAT_VECTOR, dim=384)
    schema.add_field("product_name", DataType.VARCHAR, max_length=200)

    try:
        client.create_collection(
            collection_name=collection_name, schema=schema, num_partitions=32
        )
        print(f"  ✓ Created shared collection with partition keys: {collection_name}")

        # Insert sample data for each tenant
        all_data = []
        for tenant_id in tenant_list:
            tenant_data = [
                {
                    "tenant_id": tenant_id,
                    "vector": [random.random() for _ in range(384)],
                    "product_name": f"Product {i} for {tenant_id}",
                }
                for i in range(50)
            ]
            all_data.extend(tenant_data)

        client.insert(collection_name, all_data)
        print(f"  ✓ Inserted data for {len(tenant_list)} standard tenants")

    except Exception as e:
        print(f"  ⚠️  Standard tier setup: {str(e)[:50]}...")


def setup_basic_tier(tenant_list):
    """Setup manual partition-based isolation for basic customers"""
    print("\n🏪 Setting up Basic Tier (Manual Partitions):")

    collection_name = "basic_products_manual"

    # Create collection
    schema = client.create_schema(auto_id=True)
    schema.add_field("id", DataType.INT64, is_primary=True)
    schema.add_field("tenant_id", DataType.VARCHAR, max_length=50)
    schema.add_field("vector", DataType.FLOAT_VECTOR, dim=384)
    schema.add_field("item_name", DataType.VARCHAR, max_length=200)

    try:
        client.create_collection(collection_name=collection_name, schema=schema)
        print(f"  ✓ Created basic collection: {collection_name}")

        # Create manual partitions for tenant groups
        partition_groups = [
            ["basic_user_1", "basic_user_2"],
            ["basic_user_3", "basic_user_4"],
        ]

        for i, group in enumerate(partition_groups):
            partition_name = f"basic_group_{i + 1}"
            client.create_partition(collection_name, partition_name)
            print(f"  ✓ Created partition {partition_name} for tenants: {group}")

            # Insert data for this group
            group_data = []
            for tenant_id in group:
                tenant_data = [
                    {
                        "tenant_id": tenant_id,
                        "vector": [random.random() for _ in range(384)],
                        "item_name": f"Item {j} for {tenant_id}",
                    }
                    for j in range(25)
                ]
                group_data.extend(tenant_data)

            client.insert(collection_name, group_data, partition_name)

        print("  ✓ Populated basic tier with sample data")

    except Exception as e:
        print(f"  ⚠️  Basic tier setup: {str(e)[:50]}...")


# Implement the hybrid strategy
hybrid_config = implement_hybrid_strategy()

# Setup each tier
setup_enterprise_tier(hybrid_config["enterprise_tenants"]["tenants"])
setup_standard_tier(hybrid_config["standard_tenants"]["tenants"])
setup_basic_tier(hybrid_config["basic_tenants"]["tenants"])
```

## Cross-Tier Performance Testing

```python
def test_hybrid_performance():
    """Test performance across different tiers in hybrid setup"""
    print("\n⚡ Hybrid Strategy Performance Testing")
    print("=" * 50)

    test_vector = [random.random() for _ in range(384)]
    results = {}

    # Test Enterprise Tier (Database isolation)
    print("\n🏢 Testing Enterprise Tier:")
    try:
        # Switch to enterprise database
        client.using_database("db_enterprise_corp_a")

        # Create index and load if needed
        try:
            client.create_index(
                "enterprise_products",
                "vector",
                {"index_type": "FLAT", "metric_type": "COSINE"},
            )
            client.load_collection("enterprise_products")
        except Exception:
            pass  # May already exist

        start_time = time.time()
        client.search(
            collection_name="enterprise_products", data=[test_vector], limit=5
        )
        enterprise_time = (time.time() - start_time) * 1000
        results["enterprise"] = enterprise_time
        print(f"  Enterprise query time: {enterprise_time:.2f}ms")

    except Exception as e:
        print(f"  ⚠️  Enterprise test failed: {str(e)[:40]}...")
        results["enterprise"] = None
    finally:
        client.using_database("default")

    # Test Standard Tier (Partition keys)
    print("\n🏬 Testing Standard Tier:")
    try:
        # Create index and load
        try:
            client.create_index(
                "standard_products_pk",
                "vector",
                {"index_type": "FLAT", "metric_type": "COSINE"},
            )
            client.load_collection("standard_products_pk")
        except Exception:
            pass

        start_time = time.time()
        client.search(
            collection_name="standard_products_pk",
            data=[test_vector],
            filter='tenant_id == "standard_company_1"',
            limit=5,
        )
        standard_time = (time.time() - start_time) * 1000
        results["standard"] = standard_time
        print(f"  Standard query time: {standard_time:.2f}ms")

    except Exception as e:
        print(f"  ⚠️  Standard test failed: {str(e)[:40]}...")
        results["standard"] = None

    # Test Basic Tier (Manual partitions)
    print("\n🏪 Testing Basic Tier:")
    try:
        # Create index and load
        try:
            client.create_index(
                "basic_products_manual",
                "vector",
                {"index_type": "FLAT", "metric_type": "COSINE"},
            )
            client.load_collection("basic_products_manual")
        except Exception:
            pass

        start_time = time.time()
        client.search(
            collection_name="basic_products_manual",
            data=[test_vector],
            partition_names=["basic_group_1"],
            filter='tenant_id == "basic_user_1"',
            limit=5,
        )
        basic_time = (time.time() - start_time) * 1000
        results["basic"] = basic_time
        print(f"  Basic query time: {basic_time:.2f}ms")

    except Exception as e:
        print(f"  ⚠️  Basic test failed: {str(e)[:40]}...")
        results["basic"] = None

    # Performance summary
    print("\n📊 Performance Summary:")
    valid_results = {k: v for k, v in results.items() if v is not None}

    if valid_results:
        fastest = min(valid_results.values())
        slowest = max(valid_results.values())

        for tier, time_ms in valid_results.items():
            relative = (
                "(fastest)"
                if time_ms == fastest
                else f"({time_ms / fastest:.1f}x slower)"
            )
            print(f"  {tier.title()}: {time_ms:.2f}ms {relative}")

        print(f"\n  Performance range: {fastest:.2f}ms - {slowest:.2f}ms")
        print(f"  Variance: {((slowest - fastest) / fastest) * 100:.1f}%")

    return results


# Run performance tests
perf_results = test_hybrid_performance()
```

## Cost and Resource Analysis

```python
def analyze_hybrid_costs():
    """Analyze costs and resource usage of hybrid approach"""
    print("\n💰 Hybrid Strategy Cost Analysis")
    print("=" * 50)

    # Define cost factors (relative to baseline)
    cost_factors = {
        "enterprise": {
            "storage_multiplier": 1.5,  # Database overhead
            "compute_multiplier": 1.4,  # Dedicated resources
            "operational_multiplier": 3.0,  # High maintenance
            "tenant_count": 2,
        },
        "standard": {
            "storage_multiplier": 0.3,  # Shared storage
            "compute_multiplier": 0.4,  # Shared compute
            "operational_multiplier": 0.5,  # Automated
            "tenant_count": 3,
        },
        "basic": {
            "storage_multiplier": 0.2,  # Highly shared
            "compute_multiplier": 0.3,  # Highly shared
            "operational_multiplier": 0.8,  # Manual management
            "tenant_count": 4,
        },
    }

    baseline_cost_per_tenant = 100  # Base monthly cost per tenant

    total_cost = 0
    total_tenants = 0

    print("Cost breakdown by tier:")

    for tier, factors in cost_factors.items():
        tenant_count = factors["tenant_count"]

        storage_cost = (
            baseline_cost_per_tenant * factors["storage_multiplier"] * tenant_count
        )
        compute_cost = (
            baseline_cost_per_tenant * factors["compute_multiplier"] * tenant_count
        )
        operational_cost = (
            baseline_cost_per_tenant * factors["operational_multiplier"] * tenant_count
        )

        tier_total = storage_cost + compute_cost + operational_cost
        cost_per_tenant = tier_total / tenant_count

        total_cost += tier_total
        total_tenants += tenant_count

        print(f"\n  {tier.title()} Tier ({tenant_count} tenants):")
        print(f"    Storage: ${storage_cost:.0f}/month")
        print(f"    Compute: ${compute_cost:.0f}/month")
        print(f"    Operations: ${operational_cost:.0f}/month")
        print(f"    Total: ${tier_total:.0f}/month")
        print(f"    Cost per tenant: ${cost_per_tenant:.0f}/month")

    avg_cost_per_tenant = total_cost / total_tenants

    print("\n📊 Summary:")
    print(f"  Total monthly cost: ${total_cost:.0f}")
    print(f"  Total tenants: {total_tenants}")
    print(f"  Average cost per tenant: ${avg_cost_per_tenant:.0f}")

    # Compare to single-strategy approaches
    print("\n🔍 Comparison to single strategies:")

    single_strategies = {
        "All Database": total_tenants * baseline_cost_per_tenant * 2.5,
        "All Partition Key": total_tenants * baseline_cost_per_tenant * 0.4,
        "All Manual Partition": total_tenants * baseline_cost_per_tenant * 0.6,
    }

    for strategy, cost in single_strategies.items():
        savings = ((cost - total_cost) / cost) * 100
        comparison = "savings" if savings > 0 else "additional cost"
        print(f"  vs {strategy}: ${cost:.0f} ({abs(savings):.1f}% {comparison})")

    return {
        "total_cost": total_cost,
        "cost_per_tenant": avg_cost_per_tenant,
        "tenant_count": total_tenants,
    }


# Analyze costs
cost_analysis = analyze_hybrid_costs()
```
